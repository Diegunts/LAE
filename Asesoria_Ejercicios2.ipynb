```python
import numpy as np
rng = np.random.default_rng(42)
```

1.- Dos equipos juegan una serie de juegos (4 de 7, el que gane 4 juegos gana). Cada equipo tiene un 50% de probabilidad de ganar cada juego (no empates). ¿Cuál es la probabilidad de que la serie se vaya a 7 juegos? (5/16)

```python
N = 1000000

# Se simulan 7 juegos y si llega a 4 vistorias un equipo se termina. 
juegos = rng.integers(0, 2, size=(N, 7))   # 0 = gana A, 1 = gana B

T7 = 0 #Contador de juegos que llegan a 7 partidos
for serie in juegos:
    wins_a = wins_b = 0 #Los empezamos en 0
    for j, resultado in enumerate(serie):
        if resultado == 0:
            wins_a += 1
        else:
            wins_b += 1
        if wins_a == 4 or wins_b == 4:
            if j == 6:          # se necesitaron los 7 juegos
                T7 += 1
            break

prob = T7 / N
print(f"Probabilidad simulada : {prob:.4f}") #Frecuentemente es un aproximado a 5/16 = 0.3125
```

2.- Una en mil personas tienen una enfermedad H, y la prueba de la enfermedad sale correcta 98% de las veces cuando la persona está enferma. Por otro lado, si la persona no está enferma la prueba tiene un error del 1%. Si alguien se hace una prueba y sale positiva, ¿cuál es la probabilidad de que tengan la enfermedad? (98/1097)

```python
N = 1000000

enfermos= rng.random(N) < 0.001          # 1 en 1000 tiene la enfermedad

# Resultado de la prueba para cada persona
prueba_pos = np.where(
    enfermos,
    rng.random(N) < 0.98,   # sensibilidad: 98%
    rng.random(N) < 0.01    # tasa de falso positivo: 1%
)

positivos = prueba_pos.sum()
verdaderos_pos= (enfermos&prueba_pos).sum()

prob = verdaderos_pos / positivos
print(f"Probabilidad simulada  : {prob:.4f}")
#el valor exacto de 98/1097 es .08933
```

5.- Tienes diez dados justos. Si arrojas todos al mismo tiempo, ¿cuál es la probabilidad de que la suma de los dados sea múltiplo de 6?

```python
n = 1000000

dados  = rng.integers(1, 7, size=(n, 10))
sumas  = dados.sum(axis=1) #aqui se suman los valores de cada fila.

prob = (sumas % 6 == 0).mean()
print(f"Probabilidad simulada : {prob:.4f}")
#es un aproximado a 1/6 = 0.16666, debido a que siempre el ultimo dado es el que determina
#  si la suma es múltiplo de 6 o no.
```
