```python
import pandas as pd
import statsmodels.api as sm


df = pd.read_excel("Aguas.xlsx")
print(df)
```

## Convertir los datos a booleanos  


```python
# Convertir variables categóricas a variables dummy (0 y 1)
df['Tamaño_Bool'] = df['Tamaño'].map({'Grande': 1, 'Chica': 0})
df['Hielo_Bool'] = df['Hielo'].map({'Con': 1, 'Sin': 0})
df['Kahlua'] = df['Sabor'].map({'Kahlua': 1, 'Horchata': 0})
df['Horchata'] = df['Sabor'].map({'Horchata': 1, 'Kahlua': 0})
print(df)
```

## Regresion con StatsModel

```python
# Usar las columnas del DataFrame para X e Y
y = df['Score']
x = df[['Tamaño_Bool', 'Hielo_Bool', 'Kahlua', 'Horchata']]

# Añadir constante y ajustar el modelo de regresión lineal
x_const = sm.add_constant(x)
modelo = sm.OLS(y, x_const).fit()

# Mostrar el resumen de la regresión
print(modelo.summary())
```

```python
# Convertir variables categóricas a variables dummy (-1 y 1)
df['Tamaño_Bool'] = df['Tamaño'].map({'Grande': -1, 'Chica': 1})
df['Hielo_Bool'] = df['Hielo'].map({'Con': 1, 'Sin': -1})
df['Kahlua'] = df['Sabor'].map({'Kahlua': 1, 'Horchata': 0})
df['Horchata'] = df['Sabor'].map({'Horchata': 1, 'Kahlua': 0})
print(df)
```

```python
# Ejemplo de regresión para datos de 1 y -1
import statsmodels.api as sm

# Usar una columna existente, por ejemplo Tamaño_Bool
X_bin = sm.add_constant(df[['Tamaño_Bool','Hielo_Bool','Kahlua','Horchata']])
modelo_bin = sm.OLS(y, X_bin).fit()
print(modelo_bin.summary())
```

### Analisis de las regresiones 
En una comparativa de las regresiones, no se presentan muchas diferencias en lo que son R^2, f estadistico, ni  en los P value de tamaño, hielo, kahlua y horchata, pero lo que si cambio es el beta de los mismos y tambien hubo un cambio en el intercepto, lo que nos quiere decir, los valores que se les asignan unicamente cambian la frecuencia de aumento o decremento en el modelo.
