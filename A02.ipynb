```python
import pandas as pd
import numpy as np

df = pd.read_excel("Boston Housing Dataset 1978.xlsx")
df.head()
```

## Escalamiento automático (MinMax)

```python
#Automatica 
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df_scaled = pd.DataFrame(scaler.fit_transform(df), columns=df.columns)
df_scaled.head()
```

## Manual

```python
# Normalización manual 
df_manual = df.copy()
for col in df_manual.columns:
    df_manual[col] = (df_manual[col] - df_manual[col].min()) / (df_manual[col].max() - df_manual[col].min())
df_manual.head()
```

```python
df_manual.describe()
```

## Histograma

```python
import matplotlib.pyplot as plt

hist = df.plot.hist(bins=30, figsize=(15, 10), alpha=0.5)
plt.title("Histograma de variables normalizadas (Min-Max)", fontsize=18)
plt.xlabel("Valor normalizado", fontsize=14)
plt.ylabel("Frecuencia", fontsize=14)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
