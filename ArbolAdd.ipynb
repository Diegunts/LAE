```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv('Advertising.csv', index_col=0)
print(df.shape)
df.head()
```

```python
X = df[['TV', 'radio', 'newspaper']]
y = df['sales']
```

## Dataset completo

```python
# Regresion lineal multiple — todas las variables
lr = LinearRegression()
lr.fit(X, y)

print('Intercepto:', round(lr.intercept_, 4))
for col, coef in zip(X.columns, lr.coef_):
    print(f'{col}: {round(coef, 4)}')
```

```python
# R2 regresion lineal
r2_lr = r2_score(y, lr.predict(X))
print('R2 Regresion Lineal:', round(r2_lr, 4))
```

```python
# Arbol de decision para regresion
dt = DecisionTreeRegressor(random_state=42)
dt.fit(X, y)

print('Importancia de variables:')
for col, imp in zip(X.columns, dt.feature_importances_):
    print(f'{col}: {round(imp, 4)}')
```

```python
# R2 arbol de decision
r2_dt = r2_score(y, dt.predict(X))
print('R2 Arbol de Decision:', round(r2_dt, 4))
```

## Train / Test Split 50-50

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)
print('Train:', X_train.shape[0], '| Test:', X_test.shape[0])
```

```python
# Regresion lineal con split
lr2 = LinearRegression()
lr2.fit(X_train, y_train)

print('Intercepto:', round(lr2.intercept_, 4))
for col, coef in zip(X.columns, lr2.coef_):
    print(f'{col}: {round(coef, 4)}')
```

```python
# R2 regresion lineal — evaluado en test
r2_lr2 = r2_score(y_test, lr2.predict(X_test))
print('R2 Regresion Lineal (test):', round(r2_lr2, 4))
```

```python
# Arbol de decision con split
dt2 = DecisionTreeRegressor(random_state=42)
dt2.fit(X_train, y_train)

print('Importancia de variables:')
for col, imp in zip(X.columns, dt2.feature_importances_):
    print(f'{col}: {round(imp, 4)}')
```

```python
# R2 arbol de decision — evaluado en test
r2_dt2 = r2_score(y_test, dt2.predict(X_test))
print('R2 Arbol de Decision (test):', round(r2_dt2, 4))
```

## Resumen

```python
resumen = pd.DataFrame({
    'Modelo': ['Reg. Lineal', 'Arbol Decision', 'Reg. Lineal (split)', 'Arbol Decision (split)'],
    'Evaluacion': ['Dataset completo', 'Dataset completo', 'Test 50%', 'Test 50%'],
    'R2': [round(r2_lr,4), round(r2_dt,4), round(r2_lr2,4), round(r2_dt2,4)]
})
resumen
```
